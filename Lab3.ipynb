This notebook demonstrates PatientExam class, loading patient exam data from a CSV file, calculating the average BMI, finding the busiest month, and testing error handling.

```python
from classes import PatientExam
import csv
import calendar
from collections import Counter
```

## Test PatientExam class

```python
test_exam = PatientExam(1, "12/17/2002", "Wyn Grey", 90, 1.76)

print("BMI: ", round(test_exam.get_BMI(), 2))
print("Exam Month: ", test_exam.get_exam_month())
```

## Load patient data

```python
exams = []

with open("patient_data.csv", "r", newline="") as patient_data:
    reader = csv.reader(patient_data)
    next(reader)

    for row in reader:
        exam = PatientExam(
            exam_id=int(row[0]),
            date=row[1],
            name=row[2].strip(),
            weight=int(row[3]),
            height=float(row[4])
        )
        exams.append(exam)

print("Number of patient exams have been loaded: ", len(exams))
```

## Calculate Statistics

```python
avg_bmi = sum(exam.get_BMI() for exam in exams) / len(exams)

busiest_month = calendar.month_name[
    Counter(exam.get_exam_month() for exam in exams).most_common(1)[0][0]
]

print(f"Average BMI: {avg_bmi:.2f}")
print(f"Busiest month: {busiest_month}")
```
